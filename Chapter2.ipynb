### **2.1 Attending to different parts of the input with self-attention**

### 2.1.1 A simple self-attention mechanism without trainable weights

```python
import torch


inputs = torch.tensor(
    [[0.43, 0.15, 0.89], # Yours (x^1)
     [0.55, 0.87, 0.66], # journey (x^2)
     [0.57, 0.85, 0.64], # starts (x^3)
     [0.22, 0.58, 0.33], # with (x^4)
     [0.77, 0.25, 0.10], # one (x^5)
     [0.05, 0.80, 0.55]] # step (x^6)
)
```

```python
input_query = inputs[1]
input_query
```

```python
input_1 = inputs[0]
input_1
```

```python
# lets do the dot product between input_query and input_1 using PyTorch

torch.dot(input_query, input_1)
```
